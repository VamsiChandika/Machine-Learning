```python
# Train Test Split
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
# Loading CSV file data from kaggle
data = pd.read_csv(r"C:\Users\Vamsi\Desktop\AIML\Datasets\Housing.csv")
data.head()
```

```python
data.describe()
```

```python
# Training and Testing -- (80% & 20%)
x = data[['area',]]
y = data['price']
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.xlabel('Area')
plt.ylabel('Price')
plt.title('House Price Vs Area')
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
```

```python
predicted_results = model.predict(x)
predicted_results
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y, color='blue', label="Training data points")
plt.scatter(x['area'],model.predict(x), color='red', marker='*', label = 'predicted_results')
plt.plot(x,model.predict(x), color="black", label=" linear regression")
plt.xlabel('Area')
plt.ylabel('Price')
plt.title('House Price Vs Area')
plt.show()
```

```python
# Train Test Split - ( 80% - 20% )
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
# Random_state - To preserve result even after executing multiple times 

print(" total number of records in x_train :", len(x_train))
print(" total number of records in x_test :", len(x_test))
print(" total number of records in y_train :", len(y_train))
print(" total number of records in y_test :", len(y_test))
```

```python
x_train, x_test, y_train, y_test = 100,200,300,400
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train) # While training we must give input & output features
```

```python
predicted_results = model.predict(x_test)
predicted_results
```

```python
predicted_value = model.predict(x_train)
predicted_value
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_train,y_train, color='red', label='Training data points')
plt.scatter(x_test,predicted_results, color='blue', label='predicted data points')
plt.plot(x_train,predicted_value, color='black', label='Linear Regression')
plt.xlabel('Area')
plt.ylabel('Price')
plt.title('House Price Vs Area')
plt.show()
```

```python
# Mathematical Formula for linear regression is :
# ( y = mx+c )
# m is the slope
# c is the intercept
# y is predicted datapoints ( output features )
# x is actual datapoints ( input features )
# This formula is used to darw straight line

model.coef_
model.intercept_
```

```python
# Importing Numpy & Pandas

import numpy as np
import pandas as pd

# 1. Creating a sample messy DataFrame

raw_data = {
    "Agent_ID": [101, 102, 103, 104, 105, 101, 106],
    "Model_Name": ["Claude-3.5","GPT-4o","Llama-3","Claude-3.5","GPT-4o","Claude-3.5",None,],
    "Tokens_Processed": [1200, 1500, np.nan, 800, 2100, 1200, 500],
    "Execution_Time_Sec": [1.2, 1.5, 0.9, 2.1, 1.8, 1.2, 3.5],
    "Status": ["Success","Success","Failed","Success","Success","Success","Success"],
}
```

```python
df = pd.DataFrame(raw_data)
df
```

```python
#Step-by-Step Data Analysis Workflow
#Here is how you process, clean, and analyze this data using essential 
#pandas methods:

# --- STEP 1 : INITIAL INSPECTION ---

print("--- Shape of Data ---")
print(f"Data Shape : {df.shape}")

print("\n--- Summary Statistics ---")
print(df.describe)
print(df.describe(include="all"))

print("\n--- Missing Values Count ---")
print(df.isnull().sum())
```

```python
# --- STEP 2: DATA CLEANING --
# Drop exact duplicate rows (Row index 5 is a duplicate of index 0)

print("--- Duplicates ---")
df = df.drop_duplicates()
df

# Fill missing numerical values with the column mean/average

print("--- Fill missing numerical values with the column mean/average ---")
df["Tokens_Processed"] = df["Tokens_Processed"].fillna ( df["Tokens_Processed"].mean())
df

# Drop rows where important categorical information is missing

print("--- Drop rows where important categorical information is missing ---")
df = df.dropna(subset=["Model_Name"])
df
```

```python
x = df[["Tokens_Processed","Execution_Time_Sec"]]
y = df["Tokens_Processed"]
print(f"Type of x : {type(x)}") # Represents input columns & always 2D ==> "Dataframes"
print(f"Type of y : {type(y)}") # Represents output columns & always 1D ==> "Series"
```

```python
data = {
    "subscription_years" : [1,2,0,1,3,0,2,1,0,3],
    "monthly_bill" : [50,80,90,45,110,95,60,85,70,55],
    "complaints" : [1,0,4,1,2,5,0,3,2,1],
    "quit" : [0,0,1,0,0,1,0,1,1,0]
    }
```

```python
import pandas as pd
data = pd.DataFrame(data)
data
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
sns.scatterplot(data=data,x="subscription_years",y="monthly_bill",size="complaints",hue="quit",palette="coolwarm",sizes=(20,200))
plt.show()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
```

```python
x=data[['subscription_years','monthly_bill','complaints']]
y=data['quit']
```

```python
X_train,X_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
model = DecisionTreeClassifier()
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```

```python
model.predict(X_test)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_csv(r'C:\Users\Vamsi\Desktop\AIML\Datasets\Housing.csv')
data.head()
```

```python
data['hotwaterheating'] = data['hotwaterheating'].map({'yes' : 1, 'no' : 0})
data['airconditioning'] = data['airconditioning'].map({'yes' : 1, 'no' : 0})
data
```

```python
x = data[['airconditioning']]
y = data['price']
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.xlabel('airconditioning')
plt.ylabel('price')
plt.title('Airconditioning Vs Price')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.scatter(x['airconditioning'],model.predict(x), marker = '*', color = 'green', label='data points')
plt.plot(x,model.predict(x), color = 'red', label = 'Linear Regression' )
plt.xlabel('airconditioning')
plt.ylabel('price')
plt.title('Airconditioning Vs Price')
plt.show()
```

```python
model.score(x,y)
```

```python
# Train Test Split - ( 80% - 20% )
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
# Random_state - To preserve result even after executing multiple times 

print(" total number of records in x_train :", len(x_train))
print(" total number of records in x_test :", len(x_test))
print(" total number of records in y_train :", len(y_train))
print(" total number of records in y_test :", len(y_test))
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train,y_train)
```

```python
x_test
```

```python
import matplotlib.pyplot as plt
plt.scatter(x_train,y_train, label = 'Actual Datapoints')
plt.scatter(x_test,model.predict(x_test), color = 'red', marker= '*', label = "Predicted Datapoints")
plt.plot(x_train,model.predict(x_train), color = 'black', label='LinearRegression')
plt.legend()
```

```python
model.score(x_test,y_test)
```

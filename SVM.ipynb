```python
from sklearn import svm
import numpy as np
```

```python
data = {
    "Weight" : np.array([[150,7],[200,8],[250,9],[300,6],[350,5],[400,4]]),
    "Fruit" : np.array([0,0,0,1,1,1])
}
```

```python
X = np.array([[150,7],[200,8],[250,9],[300,6],[350,5],[400,4]])
y = np.array([0,0,0,1,1,1])
```

```python
print(X[y==0][:,0])
print(X[y==1][:,0])
```

```python
print(X[y==0][:,1])
print(X[y==1][:,1])
```

```python
import matplotlib.pyplot as plt

plt.scatter(X[y==0][:,0],X[y==0][:,1],color='red',label="Apple")
plt.scatter(X[y==1][:,0],X[y==1][:,1],color="orange",label="Orange")
plt.xlabel("Weight")
plt.ylabel("Color Rating")
plt.legend()
plt.show()
```

```python
model = svm.SVC(kernel='rbf', gamma='scale', C=1.0)
model.fit(X,y)
```

```python
model.predict([[200,9]])
```

```python
model.score(X,y)
```

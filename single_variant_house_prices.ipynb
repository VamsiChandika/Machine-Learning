```python
data = {
    "Area" : [1000,1500,2000,2500,3000,3500],
    "Price" : [100000,300000,200000,400000,500000,150000]
}
```

```python
import pandas as pd
df = pd.DataFrame(data)
df
```

```python
x = df[["Area"]]
y = df["Price"]
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.xlabel("Area sqmts")
plt.ylabel("Price")
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.plot(x,model.predict(x),color = "red", label = "Regression Line")
plt.xlabel("Area sqmts")
plt.ylabel("Price")
plt.show()
```

```python
test_dataset = pd.DataFrame({"Area" : [1250,1750,2250,2750,3250,3750]})
test_dataset
```

```python
predicted_price = model.predict(test_dataset)
predicted_price
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.scatter(test_dataset,predicted_price,color = "green", label = "predicted_price", marker = "*")
plt.plot(test_dataset,model.predict(test_dataset),color = "red", label = "Regression Line")
plt.xlabel("Area sqmts")
plt.ylabel("Price")
plt.show()
```
